# Workable dataset — EDA starting point
Run `voc-ml dataset init` and `voc-ml dataset run` in the runtime shell first.
This notebook reads the latest **published** snapshot; it does not update stores.

```python
from voc import collector as co
snapshot = co.get_dataset(stage="prepared", version="latest")
snapshot.info
```

```python
df = snapshot.data
assert df.record_id.is_unique
df.groupby("platform").size().rename("reviews")
```

```python
from voc.datasets import summary
summary(snapshot)["month_counts"]
```

To repeat an analysis later, replace `latest` with `snapshot.info["artifact_id"]`.
Embeddings, sentiment scores, clustering and further text cleaning are separate future outputs.

```python
pinned = co.get_dataset(version=snapshot.info["artifact_id"])
assert pinned.info == snapshot.info
assert pinned.data.equals(df)
```
