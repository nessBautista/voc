# Workable dataset — EDA starting point
Use the configured dataset source (`local` or `s3`). Shared readers do not initialize a raw store or run the publisher pipeline.
This notebook reads a published snapshot; it does not collect or share data.

```python
from voc import collector as co
snapshot = co.get_dataset(stage="prepared", version="latest")
snapshot.info
```

```python
df = snapshot.data
assert df.record_id.is_unique
df.groupby("platform").size().rename("reviews")
```

```python
from voc.datasets import summary
summary(snapshot)["month_counts"]
```

Pin shared data by `release_id`; pin personal data by `artifact_id`.
Keep the source explicit when repeating an analysis. Embeddings and sentiment outputs remain separate.

```python
source = snapshot.info.get("source", "local")
version = snapshot.info.get("release_id", snapshot.info["artifact_id"])
pinned = co.get_dataset(source=source, version=version)
assert pinned.info == snapshot.info
assert pinned.data.equals(df)
```
