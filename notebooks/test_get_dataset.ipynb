```python
from voc import collector as co

snapshot = co.get_dataset(source="s3")
release_id = snapshot.info["release_id"]

raw = co.get_dataset(
    source="s3", stage="raw", version=release_id
)
pinned = co.get_dataset(source="s3", version=release_id)

assert pinned.data.equals(snapshot.data)
assert raw.info["raw_revision_id"] == snapshot.info["raw_revision_id"]

print("Release:", release_id)
print("Raw:", raw.data.shape)
print("Workable:", snapshot.data.shape)
display(snapshot.data.head())
```
